```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import bioframe
```

```python
cCRE_data = pd.read_csv('/sharehome/dlafonta/data/K562_data/cCREs/ENCFF414OGC_ENCFF806YEZ_ENCFF849TDM_ENCFF736UDR.bed',sep='\t', header=None)
column_names = ['chrom', 'start', 'end', 'cCRE_code','zeros','dot','start_x','end_x','two_twenty_fives','call','classification'] # Replace with your actual column names
cCRE_data.columns = column_names
```
